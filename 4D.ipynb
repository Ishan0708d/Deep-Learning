```python
import pandas as pd
import numpy as np
import re
from gensim.models import Word2Vec
from sklearn.metrics.pairwise import cosine_similarity

# 1. Data collection and preprocessing
data = pd.read_csv("twitter_dataset.csv")

def preprocess(text):
    text = text.lower()
    text = re.sub(r"http\S+|www\S+", "", text)
    text = re.sub(r"[^a-z\s]", "", text)
    return text.split()

tweets = data["Text"].dropna().tolist()
sentences = [preprocess(t) for t in tweets]

# 2a. Train CBOW Word2Vec model
model = Word2Vec(
    sentences=sentences,
    vector_size=100,
    window=5,
    min_count=2,
    sg=0,
    epochs=10
)

# 2b. Convert each tweet into an average embedding
def get_vector(words):
    vectors = [model.wv[w] for w in words if w in model.wv]
    return np.mean(vectors, axis=0) if vectors else np.zeros(100)

doc_vectors = np.array([get_vector(s) for s in sentences])

# 2c and 3. Recommend similar tweets
def recommend(query):
    words = preprocess(query)
    query_vector = get_vector(words)

    if not np.any(query_vector):
        print("No matching words found in vocabulary.")
        return

    scores = cosine_similarity([query_vector], doc_vectors)[0]
    indices = np.argsort(scores)[::-1][:5]

    for i in indices:
        print("\nTweet:", tweets[i])
        print("Similarity:", round(scores[i], 3))

# Example
recommend("technology and artificial intelligence")
```
