```python
import re
from gensim.models import Word2Vec

# a. Data preparation
with open("document3.txt", "r") as f:
    text = f.read()

text = text.lower()
text = re.sub(r"[^a-z\s]", "", text)
sentences = [line.split() for line in text.splitlines() if line.strip()]

# b. Train Word2Vec CBOW model
model = Word2Vec(
    sentences=sentences,
    vector_size=50,
    window=2,
    min_count=1,
    sg=0,
    epochs=100
)

# c. Evaluate results
print("Vocabulary:", list(model.wv.index_to_key))

word = "influenza"

print("Word embedding:", model.wv[word])
print("Similar words:", model.wv.most_similar(word, topn=5))
```
